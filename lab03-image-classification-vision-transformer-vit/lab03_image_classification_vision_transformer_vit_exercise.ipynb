```python
# Import Libraries
import os
import numpy as np
from PIL import Image
import keras_hub

# Load ImageNet class names
#with open("../imagenet_classes.txt") as f:
#   class_names = [line.strip() for line in f]
import urllib.request

url = "https://raw.githubusercontent.com/pytorch/hub/master/imagenet_classes.txt"
with urllib.request.urlopen(url) as f:
    class_names = [line.decode("utf-8").strip() for line in f]

# Load ViTImageClassifierPreprocessor (ImageNet preset)
preprocessor = keras_hub.models.ViTImageClassifierPreprocessor.from_preset(
    "vit_base_patch16_224_imagenet"
)

# Load ViTImageClassifier (ImageNet pretrained model)
model = keras_hub.models.ViTImageClassifier.from_preset(
    "vit_base_patch16_224_imagenet"
)

# Image folder
image_folder = "images"

# Images and true labels
images = {
    "blow dryer.jpg": "blow dryer",
    "espresso maker.jpg": "espresso maker",
    "necklace.png": "necklace",
    "pomeranian.jpg": "pomeranian",
    "sunglasses.jpg": "sunglasses"
}

print("Image File | Predicted Label | True Label | Correct?")
print("-"*65)

for file, true_label in images.items():

    path = os.path.join(image_folder, file)

    # Load image
    img = Image.open(path).convert("RGB")
    img = img.resize((224, 224))
    img = np.array(img)

    # Add batch dimension
    x = np.expand_dims(img, axis=0)

    # Predict
    preds = model.predict(x, verbose=0)

    # Get predicted class
    pred_index = np.argmax(preds[0])
    predicted_label = class_names[pred_index]

    # Check correctness
    correct = predicted_label.lower() == true_label.lower()

    print(f"{file} | {predicted_label} | {true_label} | {correct}")
```
